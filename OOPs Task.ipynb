```python
'''Q1, Create a vehicle class with an init method having instance variables as name_of_vehicle, max_speed
and average_of_vehicle.'''
```

```python
class vehicle:
    def __init__(self, name_of_vehicle, max_speed, average_of_vehicle):
        
        self.name_of_vehicle = name_of_vehicle
        self.max_speed = max_speed
        self.average_of_vehicle = average_of_vehicle

    def vehicle_details(self):
        return self.name_of_vehicle, self.max_speed, self.average_of_vehicle
```

```python
car = vehicle('Lambogini', '300 km/h', '80 km/h')
```

```python
car.average_of_vehicle
```

```python
car.vehicle_details()
```

```python
'''Q2. Create a child class car from the vehicle class created in Que 1, which will inherit the vehicle class.
Create a method named seating_capacity which takes capacity as an argument and returns the name of
the vehicle and its seating capacity.'''
```

```python
class car(vehicle):
    
    def __init__(self, name_of_vehicle, seats_capacity):

        self.name_of_vehicle = name_of_vehicle
        self.seats_capacity = seats_capacity

    def seating_capacity(self):
        return self.name_of_vehicle, self.seats_capacity

    
```

```python
car_var = car('lambogini', 2)
```

```python
car_var.name_of_vehicle
```

```python
car_var.seating_capacity()
```

```python
'''Q3. What is multiple inheritance? Write a python code to demonstrate multiple inheritance.

Multiple Inheritance

Multiple inheritance is a feature in object-oriented programming where a class can inherit attributes and methods from multiple parent classes.
It allows you to combine the characteristics of several different classes into a single new class. Think of it like a family tree where a child
can inherit traits from both their mother and father.
'''

class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print("Generic animal sound")

class Flyer:
    def fly(self):
        print("Flying...")

class Bird(Animal, Flyer):  # Inherits from both Animal and Flyer
    def __init__(self, name, color):
        super().__init__(name)  # Initialize Animal's name
        self.color = color

    def speak(self):  # Overriding the speak method
        print("Tweet tweet!")

# Example Usage
my_bird = Bird("Tweety", "Yellow")
my_bird.speak()  # Output: Tweet tweet!
my_bird.fly()  # Output: Flying...
print(my_bird.name)  # Output: Tweety
```

```python
'''Q4. What are getter and setter in python? Create a class and create a getter and a setter method in
thisclass.
Getters and Setters in Python:
Getters and setters are methods used to control access to an object's attributes. They provide a way to
encapsulate data and manage how it's accessed and modified.'''

class Person:
    def __init__(self, name, age):
        self._name = name  # Using a leading underscore indicates a protected attribute
        self._age = age

    def get_age(self):
        return self._age

    def set_age(self, new_age):
        if new_age >= 0:
            self._age = new_age
        else:
            print("Age cannot be negative.")

person = Person("Alice", 30)
print(person.get_age())  # Accessing age using the getter
person.set_age(35)       # Modifying age using the setter
print(person.get_age())
person.set_age(-5)      # Attempting to set a negative age
print(person.get_age())
```

```python
'''Q5.What is method overriding in python? Write a python code to demonstrate method overriding.

Method Overriding in Python
Method overriding is a fundamental concept in object-oriented programming (OOP) where a subclass provides
its own implementation of a method that is already defined in its superclass (parent class). This allows
subclasses to customize behavior inherited from their parent class.'''
class House:
    def describe(self):
        print("This is a house.")

class ModernHouse(House):
    def describe(self):
        print("This is a modern house with sleek lines and big windows.")

class Cottage(House):
    def describe(self):
        print("This is a cozy cottage with a charming fireplace.")

# Let's build some houses!
modern_home = ModernHouse()
cottage = Cottage()

modern_home.describe()  # Output: This is a modern house with sleek lines and big windows.
cottage.describe()  # Output: This is a cozy cottage with a charming fireplace.
```
